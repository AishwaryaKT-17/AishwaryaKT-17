Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Load the Dataset & view the first 5 rows

```python
df = pd.read_csv("heart.csv")
df.head()
```

Check Dataset Information

```python
df.info()
```

Get statistical summary

```python
df.describe()
```

Check for missing values

```python
df.isnull().sum()
```

Age Distribution

```python
df['age'].hist(figsize=(6,4))
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

Male vs Female

```python
df['sex'].value_counts().plot(kind='bar')
plt.title("Male vs Female")
plt.show()
```

Heart Disease Count

```python
df['target'].value_counts().plot(kind='bar')
plt.title("Heart Disease")
plt.show()
```

Cholesterol Distribution

```python
df['cholestoral'].hist(figsize=(6,4))
plt.title("Cholesterol Distribution")
plt.show()
```

Correlation Heatmap

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(10,8))
plt.imshow(numeric_df.corr())
plt.colorbar()
plt.xticks(range(len(numeric_df.columns)), numeric_df.columns, rotation=90)
plt.yticks(range(len(numeric_df.columns)), numeric_df.columns)
plt.show()
```

cholestoral box plot

```python
plt.boxplot(df['cholestoral'])
plt.title("Cholestoral Boxplot")
plt.show()
```

Age vs Heart Disease

```python
pd.crosstab(df['age'], df['target']).plot(figsize=(10,5))
plt.title("Age vs Heart Disease")
plt.show()
```
